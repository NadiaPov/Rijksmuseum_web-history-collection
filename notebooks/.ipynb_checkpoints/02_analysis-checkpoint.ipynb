```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, HTML

csv_path = "C:/Users/povroznik/Documents/Research/2026/JDH/Rijksmuseum_web-history-collection/Rijksmuseum_web-history-collection/data/rijksstudio_yearend_counts.csv"
df = pd.read_csv(csv_path)

# ---- Remove 2025 (empty) ----
df = df[df["year"] != 2025].copy()

# ---- Plot (all remaining years) ----
years = df["year"].astype(int)

plt.figure(figsize=(11, 6))

plt.plot(years, df["total_artworks"],
         marker="o", label="Total artworks")
plt.plot(years, df["rijksstudios"],
         marker="o", label="Rijksstudio's")

plt.title("Figure 3. Rijksmuseum: Artworks Online and Rijksstudio's Entries")
plt.xlabel("Year")
plt.ylabel("Count")
plt.grid(True, linestyle="--", linewidth=0.5)
plt.legend()
plt.xticks(years, rotation=45)
plt.tight_layout()
plt.show()

# ---------- Clean formatting ----------
def fmt_int(x):
    if pd.isna(x):
        return ""
    return f"{int(x)}"

def small_table(df_part):
    tdf = df_part[["year", "total_artworks", "rijksstudios"]].copy()
    return (
        tdf.style
           .format({
               "year": lambda x: f"{int(x)}",
               "total_artworks": fmt_int,
               "rijksstudios": fmt_int
           })
           .set_properties(**{
               "text-align": "center",
               "font-size": "11pt",
               "border": "1px solid #cccccc",
               "padding": "4px"
           })
           .set_table_styles([
               dict(selector="th", props=[
                   ("font-size", "11pt"),
                   ("text-align", "center"),
                   ("border", "1px solid #cccccc"),
                   ("padding", "4px"),
               ])
           ])
           .hide(axis="index")
           .to_html()
    )

# ---- Four fragments ----
t1 = small_table(df[(df["year"] >= 2013) & (df["year"] <= 2015)])
t2 = small_table(df[(df["year"] >= 2016) & (df["year"] <= 2018)])
t3 = small_table(df[(df["year"] >= 2019) & (df["year"] <= 2021)])
t4 = small_table(df[(df["year"] >= 2021) & (df["year"] <= 2024)])

# ---- Show all four tables horizontally ----
html = f"""
<div style="display: flex; gap: 20px;">
    <div>{t1}</div>
    <div>{t2}</div>
    <div>{t3}</div>
    <div>{t4}</div>
</div>
"""

display(HTML(html))
```

### Extracting file formats

```python
# Formats from 'original' - URL string

import pandas as pd
import os
import re

# Input and output paths
input_path = r'data/Rijksmuseum_unique_urls_per_year.csv'
output_path = r'data/Rijksmuseum_file_formats_per_year_sorted.csv'

# Load data
if not os.path.exists(input_path):
    raise FileNotFoundError(f"❌ File not found: {input_path}")

df = pd.read_csv(input_path)
print(f"✅ Loaded {len(df)} rows with columns: {list(df.columns)}")

# Check required columns
if 'original' not in df.columns or 'year' not in df.columns:
    raise ValueError("❌ Required columns 'original' and 'year' are missing.")

# Define format detection logic
def detect_format(url):
    if not isinstance(url, str):
        return 'none'
    if url.endswith('/'):
        return 'html'
    
    last_segment = url.split('/')[-1]
    
    # If there's a file extension
    if '.' in last_segment:
        match = re.search(r'\.([a-zA-Z0-9]+)$', last_segment)
        if match:
            return match.group(1).lower()
    
    # If no extension but letters/numbers present
    if re.fullmatch(r'[a-zA-Z0-9_-]+', last_segment):
        return 'html'
    
    return 'none'

# Apply the function
df['format'] = df['original'].apply(detect_format)

# Diagnostics
print("🔍 Top detected formats:")
print(df['format'].value_counts().head())

# Group by format and year
format_counts = df.groupby(['format', 'year']).size().unstack(fill_value=0)

# Sort by total across all years
format_counts['total'] = format_counts.sum(axis=1)
format_counts_sorted = format_counts.sort_values(by='total', ascending=False).drop(columns='total')

# Save to CSV
format_counts_sorted.to_csv(output_path)
print(f"✅ Format count table saved to: {output_path}")
print(f"📊 Table shape: {format_counts_sorted.shape}")
print(format_counts_sorted.head())

```

### Creating lists of HTML and Image URLs

```python
# Lists html and image
import pandas as pd
import re

# Input and output paths
input_path = r'data/Rijksmuseum_unique_urls_per_year.csv'
output_html = r'data/Rijksmuseum_urls_html_formats.csv'
output_images = r'data/Rijksmuseum_urls_image_formats.csv'

# Load data
df = pd.read_csv(input_path, low_memory=False)

# Extract file format from URL
def extract_format(url):
    if isinstance(url, str):
        if url.endswith('/'):
            return '[slash]'
        match = re.search(r'\.([a-zA-Z0-9]+)(?:[\?#]|$)', url)
        if match:
            return match.group(1).lower()
    return '[none]'

df['format'] = df['original'].apply(extract_format)

# Define format categories
html_formats = {'[none]', 'htm', 'html'}
image_formats = {'gif', 'jpg', 'jpeg', 'png', 'imageurl'}

# Filter by format
html_df = df[df['format'].isin(html_formats)]
image_df = df[df['format'].isin(image_formats)]

# Save to CSV
html_df.to_csv(output_html, index=False)
image_df.to_csv(output_images, index=False)

# Summary
print(f"✅ HTML-like URLs saved to: {output_html}")
print(f"✅ Image URLs saved to: {output_images}")
print(f"📊 HTML rows: {len(html_df)} | Image rows: {len(image_df)}")

```

### URLs segmentation (HTML only)
During segmentation, the ‘/’ sign has been used as a segment identifier. 

```python
# URLs segmentation into parts. HTML only

import pandas as pd
from urllib.parse import urlparse
from collections import defaultdict
import re

# Input and output paths (local to your machine)
input_path = r'data/Rijksmuseum_urls_html_formats.csv'
output_path = r'data/url_segment_counts_by_year_HTML_only.csv'

# Load data
df = pd.read_csv(input_path, low_memory=False)

# Extract year
if 'timestamp' in df.columns:
    df['year'] = df['timestamp'].astype(str).str[:4]

# Helper: extract path segments from a Rijksmuseum URL
def extract_segments(url):
    try:
        parsed = urlparse(url)
        if 'rijksmuseum.nl' not in parsed.netloc:
            return []
        path = parsed.path.strip('/')
        return path.split('/')
    except Exception:
        return []

# Count occurrences per segment position and year
segment_counts = defaultdict(lambda: defaultdict(int))

for _, row in df.iterrows():
    url = row.get('url') or row.get('original')
    year = str(row.get('year'))
    if pd.isna(url) or pd.isna(year):
        continue
    segments = extract_segments(url)
    for i, seg in enumerate(segments):
        col_label = f'Segment {i+1}'
        segment_counts[(col_label, seg)][year] += 1

# Convert to DataFrame
rows = []
for (seg_col, seg_val), year_data in segment_counts.items():
    row = {seg_col: seg_val}
    for year, count in year_data.items():
        row[str(year)] = count
    rows.append(row)

df_segments = pd.DataFrame(rows)

# Normalize: ensure segment columns align properly
segment_cols = sorted(
    [col for col in df_segments.columns if isinstance(col, str) and col.startswith("Segment")],
    key=lambda x: int(re.search(r'\d+', x).group())
)
year_cols = sorted([col for col in df_segments.columns if col not in segment_cols])
df_segments = df_segments[segment_cols + year_cols]

# Save to CSV
df_segments.to_csv(output_path, index=False)

print(f"✅ CSV file saved at: {output_path}")
```

### Language versioning

```python
# Extract html by language versions into separate csv files

import pandas as pd
from urllib.parse import urlparse
import os

# === Paths ===
input_path = r'data/Rijksmuseum_urls_html_formats.csv'
base_path = os.path.splitext(input_path)[0]

# === Load data ===
df = pd.read_csv(input_path)

# === Extract segments ===
def extract_segment_1(url):
    try:
        path = urlparse(url).path.strip('/')
        return path.split('/')[0] if path else None
    except:
        return None

def extract_final_segment(url):
    try:
        path = urlparse(url).path.strip('/')
        parts = path.split('/')
        if not parts:
            return None
        return parts[-1] if parts[-1] else (parts[-2] if len(parts) > 1 else None)
    except:
        return None

if 'original' not in df.columns:
    raise ValueError("❌ 'original' column is missing in the dataset.")

df['Segment 1'] = df['original'].astype(str).apply(extract_segment_1)
df['Final Segment'] = df['original'].astype(str).apply(extract_final_segment)

# === Extract year from timestamp ===
if 'timestamp' in df.columns:
    df['year'] = df['timestamp'].astype(str).str[:4]
else:
    raise ValueError("❌ 'timestamp' column is missing — cannot extract year.")

# === Filter by language ===
language_codes = ['nl', 'en', 'uk', 'de']
lang_df = df[df['Segment 1'].isin(language_codes)].copy()

# === Build lookup: final segment -> languages ===
segment_lang_map = (
    lang_df.groupby('Final Segment')['Segment 1']
    .apply(lambda x: sorted(set(x)))
    .to_dict()
)

# === Map languages per final segment ===
lang_df['Languages With Same Final Segment'] = lang_df['Final Segment'].map(
    lambda fs: ','.join(segment_lang_map.get(fs, []))
)

# === Summaries ===
summary = []
year_distribution_df = pd.DataFrame()
matched_links = []

for lang in language_codes:
    lang_data = lang_df[lang_df['Segment 1'] == lang].copy()
    total = len(lang_data)
    unique_final = lang_data['Final Segment'].nunique()

    # Check if final segment appears in other languages
    lang_data['Matched Other Lang'] = lang_data['Languages With Same Final Segment'].apply(
        lambda x: any(l != lang for l in x.split(','))
    )
    matched = lang_data['Matched Other Lang'].sum()
    matched_links.append(lang_data[lang_data['Matched Other Lang']])

    # Per-year summary
    year_summary = lang_data.groupby('year')['Matched Other Lang'].agg(['sum', 'count']).reset_index()
    year_summary.columns = ['year', 'Similar in other langs', 'Total']
    year_summary['Language'] = lang
    year_distribution_df = pd.concat([year_distribution_df, year_summary], ignore_index=True)

    summary.append({
        'Language': lang,
        'Total URLs': total,
        'Unique Final Segments': unique_final,
        'Similar Final Segments in Other Languages': matched
    })

# === Save outputs ===
# 1. Matched URLs
matched_df = pd.concat(matched_links).drop_duplicates()
matched_df.to_csv(base_path + '_matched_by_final_segment.csv', index=False)

# 2. Summary
pd.DataFrame(summary).to_csv(base_path + '_segment1_comparison_summary.csv', index=False)

# 3. Year breakdown
year_distribution_df.to_csv(base_path + '_segment1_year_distribution.csv', index=False)

# 4. Save per-language files
for lang in language_codes:
    lang_subset = lang_df[lang_df['Segment 1'] == lang]
    lang_subset.to_csv(base_path + f'_urls_lang_{lang}.csv', index=False)

# === Output info ===
print("✅ All results saved:")
print(" - Matched URLs:", base_path + '_matched_by_final_segment.csv')
print(" - Summary Stats:", base_path + '_segment1_comparison_summary.csv')
print(" - Yearly Distribution:", base_path + '_segment1_year_distribution.csv')
print(" - Per-language files:")
for lang in language_codes:
    print(f"   • {lang}: {base_path}_urls_lang_{lang}.csv")
```

### Checking overlapping files by final segment of the URL-strings in different language versionings (NL, EN, DE)

```python

import pandas as pd
from urllib.parse import urlparse
import matplotlib.pyplot as plt
from matplotlib_venn import venn3

# --- File paths ---
files = {
    "nl": r'data/Rijksmuseum_urls_html_formats_urls_lang_nl.csv',
    "en": r'data/Rijksmuseum_urls_html_formats_urls_lang_en.csv',
    "de": r'data/Rijksmuseum_urls_html_formats_urls_lang_de.csv',
}

# --- Helper to extract final segment ---
def get_final_segment(url):
    try:
        path = urlparse(url).path.strip('/')
        segments = path.split('/')
        if not segments or segments == ['']:
            return None
        return segments[-1] if segments[-1] else (segments[-2] if len(segments) > 1 else None)
    except:
        return None

# --- Read data and extract segments ---
final_segments = {}
for lang, path in files.items():
    df = pd.read_csv(path)
    if 'original' not in df.columns:
        raise ValueError(f"❌ 'original' column missing in {lang} file.")
    segments = df['original'].astype(str).apply(get_final_segment)
    final_segments[lang] = set(segments.dropna().unique())

# --- Create Venn diagram ---
plt.figure(figsize=(10, 8))
venn3(
    [final_segments['nl'], final_segments['en'], final_segments['de']],
    set_labels=('nl', 'en', 'de')
)
plt.title("Overlap of Final URL Segments across Languages")
plt.show()

# --- Print summary ---
for lang, segs in final_segments.items():
    print(f"🔹 {lang}: {len(segs)} unique final segments")

overlap_nl_en = len(final_segments['nl'] & final_segments['en'])
overlap_nl_de = len(final_segments['nl'] & final_segments['de'])
overlap_en_de = len(final_segments['en'] & final_segments['de'])
overlap_all = len(final_segments['nl'] & final_segments['en'] & final_segments['de'])

print(f"\n🔁 Overlaps:")
print(f"nl ∩ en: {overlap_nl_en}")
print(f"nl ∩ de: {overlap_nl_de}")
print(f"en ∩ de: {overlap_en_de}")
print(f"nl ∩ en ∩ de: {overlap_all}")
```

### Extracting and separating 'collection' and 'collectie' related URLs

```python
# 'Collection' and 'collectie' in the urls extracted separately
import pandas as pd
from urllib.parse import urlparse
import os

# Load data
input_path = r'data/Rijksmuseum_urls_html_formats.csv'
df = pd.read_csv(input_path)

# --- Functions to match individual keywords in segments ---
def has_collection(url):
    try:
        path = urlparse(str(url)).path.strip('/')
        segments = path.split('/')
        return any('collection' in seg.lower() for seg in segments)
    except:
        return False

def has_collectie(url):
    try:
        path = urlparse(str(url)).path.strip('/')
        segments = path.split('/')
        return any('collectie' in seg.lower() for seg in segments)
    except:
        return False

# Filter separately
df_collection = df[df['original'].apply(has_collection)]
df_collectie = df[df['original'].apply(has_collectie)]

# Output file paths
base_dir = os.path.dirname(input_path)
output_collection = os.path.join(base_dir, 'urls_with_collection.csv')
output_collectie = os.path.join(base_dir, 'urls_with_collectie.csv')

# Save to CSV
df_collection.to_csv(output_collection, index=False)
df_collectie.to_csv(output_collectie, index=False)

# Report
print(f"✅ {len(df_collection)} records saved to:\n{output_collection}")
print(f"✅ {len(df_collectie)} records saved to:\n{output_collectie}")
```

### Checking NL and EN coverage in the Internet Archive by comparing 'collection' (EN) and 'collectie' (NL) segment return

```python
# Distribution by Year for collection vs collectie

import pandas as pd
from urllib.parse import urlparse
import os

# Load data
input_path = r'data/Rijksmuseum_urls_html_formats.csv'
df = pd.read_csv(input_path)

# Function to extract year from URL (if not in a 'year' column)
def extract_year(url):
    try:
        match = pd.to_numeric(pd.Series(url).str.extract(r'/(\d{4})/')[0], errors='coerce')
        return int(match.iloc[0]) if not pd.isna(match.iloc[0]) else None
    except:
        return None

# Add 'year' column if it doesn't exist
if 'year' not in df.columns:
    df['year'] = df['original'].apply(extract_year)
    df['year'] = df['year'].fillna('Unknown')

# --- Functions to match keywords ---
def has_collection(url):
    try:
        path = urlparse(str(url)).path.strip('/')
        segments = path.split('/')
        return any('collection' in seg.lower() for seg in segments)
    except:
        return False

def has_collectie(url):
    try:
        path = urlparse(str(url)).path.strip('/')
        segments = path.split('/')
        return any('collectie' in seg.lower() for seg in segments)
    except:
        return False

# Filter datasets
df['has_collection'] = df['original'].apply(has_collection)
df['has_collectie'] = df['original'].apply(has_collectie)

# Group and count by year
collection_by_year = df[df['has_collection']].groupby('year').size().rename('collection')
collectie_by_year = df[df['has_collectie']].groupby('year').size().rename('collectie')

# Combine into one table
yearly_comparison = pd.concat([collection_by_year, collectie_by_year], axis=1).fillna(0).astype(int).sort_index()

# Save to CSV
output_path = input_path.replace('.csv', '_collection_vs_collectie_by_year.csv')
yearly_comparison.to_csv(output_path)

# Print summary
print("📊 Distribution by year:")
print(yearly_comparison)
print(f"\n✅ Saved to: {output_path}")
```

### Extracting object numbers from the URLs

```python
# Identifying urls with objects by extracting the code 

import pandas as pd
import re
import os

# --- Define file paths ---
base_dir = r'data'

files = {
    'collection': os.path.join(base_dir, 'urls_with_collection.csv'),
    'collectie': os.path.join(base_dir, 'urls_with_collectie.csv')
}

# --- Define flexible pattern for codes ---
# Match codes like: RP-P-BI-5253, BK-15400, SK-C-5, AK-MAK-1160
code_pattern = re.compile(r'\b(?:[A-Z0-9]+-){1,}[A-Z0-9]+\b')

# --- Function to check if URL contains a code ---
def contains_code(url):
    return bool(code_pattern.search(str(url)))

# --- Process both files ---
for label, path in files.items():
    print(f"\n🔍 Processing: {os.path.basename(path)}")

    # Read CSV
    df = pd.read_csv(path)

    # Check for code in each URL
    df['has_code'] = df['original'].apply(contains_code)

    # Split into with/without code
    df_with_code = df[df['has_code']].drop(columns='has_code')
    df_without_code = df[~df['has_code']].drop(columns='has_code')

    # Define output paths
    output_with = os.path.join(base_dir, f'urls_with_{label}_with_code.csv')
    output_without = os.path.join(base_dir, f'urls_with_{label}_without_code.csv')

    # Save both files
    df_with_code.to_csv(output_with, index=False)
    df_without_code.to_csv(output_without, index=False)

    # Report
    print(f"✅ Saved {len(df_with_code)} URLs WITH code → {output_with}")
    print(f"✅ Saved {len(df_without_code)} URLs WITHOUT code → {output_without}")



```

### Extracting object numbers (the code) from 'collection' and 'collectie' related URLs and comparing them in the datasets (EN vs NL)

```python
# Compare datasets by codes (first it extracts the code)

import pandas as pd
import re
import os

# Define file paths
base_dir = r'data'
file1 = os.path.join(base_dir, 'urls_with_collection_with_code.csv')
file2 = os.path.join(base_dir, 'urls_with_collectie_with_code.csv')

# Regex pattern for codes like RP-P-BI-5253, SK-C-5 etc.
code_pattern = re.compile(r'\b(?:[A-Z0-9]+-){1,}[A-Z0-9]+\b')

# Function to extract code from URL
def extract_code(url):
    match = code_pattern.search(str(url))
    return match.group(0) if match else None

# Load files and extract codes
df1 = pd.read_csv(file1)
df1['code'] = df1['original'].apply(extract_code)

df2 = pd.read_csv(file2)
df2['code'] = df2['original'].apply(extract_code)

# Drop rows where code couldn't be extracted
df1 = df1.dropna(subset=['code'])
df2 = df2.dropna(subset=['code'])

# Get sets
codes1 = set(df1['code'])
codes2 = set(df2['code'])

# Compare
common_codes = codes1 & codes2
only_in_1 = codes1 - codes2
only_in_2 = codes2 - codes1

# Filter DataFrames
df_common = pd.concat([df1[df1['code'].isin(common_codes)], df2[df2['code'].isin(common_codes)]])
df_only_1 = df1[df1['code'].isin(only_in_1)]
df_only_2 = df2[df2['code'].isin(only_in_2)]

# Output
df_common.to_csv(os.path.join(base_dir, 'codes_in_both.csv'), index=False)
df_only_1.to_csv(os.path.join(base_dir, 'codes_only_in_collection.csv'), index=False)
df_only_2.to_csv(os.path.join(base_dir, 'codes_only_in_collectie.csv'), index=False)

# Print results in numbers
print("🔢 Comparison Summary:")
print(f"📁 Total in collection: {len(codes1)}")
print(f"📁 Total in collectie: {len(codes2)}")
print(f"✅ Codes in both files: {len(common_codes)}")
print(f"➖ Codes only in collection: {len(only_in_1)}")
print(f"➖ Codes only in collectie: {len(only_in_2)}")

print("✅ Comparison completed and files saved.")
```

### Visualising overlapping objects

```python
# Visualisation of overlapping objects

import pandas as pd
import re
import os
import matplotlib.pyplot as plt
from matplotlib_venn import venn2

# === Define file paths ===
base_dir = r'data'
file1 = os.path.join(base_dir, 'urls_with_collection_with_code.csv')
file2 = os.path.join(base_dir, 'urls_with_collectie_with_code.csv')

# === Regex pattern for artwork codes ===
code_pattern = re.compile(r'\b(?:[A-Z0-9]+-){1,}[A-Z0-9]+\b')

# === Function to extract code from URL ===
def extract_code(url):
    match = code_pattern.search(str(url))
    return match.group(0) if match else None

# === Load files and extract codes ===
df1 = pd.read_csv(file1)
df1['code'] = df1['original'].apply(extract_code)

df2 = pd.read_csv(file2)
df2['code'] = df2['original'].apply(extract_code)

# === Drop rows without code ===
df1 = df1.dropna(subset=['code'])
df2 = df2.dropna(subset=['code'])

# === Sets for comparison ===
codes1 = set(df1['code'])
codes2 = set(df2['code'])

# === Compare ===
common_codes = codes1 & codes2
only_in_1 = codes1 - codes2
only_in_2 = codes2 - codes1

# === Filter DataFrames ===
df_common = pd.concat([df1[df1['code'].isin(common_codes)], df2[df2['code'].isin(common_codes)]])
df_only_1 = df1[df1['code'].isin(only_in_1)]
df_only_2 = df2[df2['code'].isin(only_in_2)]

# === Output CSVs ===
df_common.to_csv(os.path.join(base_dir, 'codes_in_both.csv'), index=False)
df_only_1.to_csv(os.path.join(base_dir, 'codes_only_in_collection.csv'), index=False)
df_only_2.to_csv(os.path.join(base_dir, 'codes_only_in_collectie.csv'), index=False)

# === Print summary ===
print("🔢 Comparison Summary:")
print(f"📁 Total in collection: {len(codes1)}")
print(f"📁 Total in collectie: {len(codes2)}")
print(f"✅ Codes in both files: {len(common_codes)}")
print(f"➖ Codes only in collection: {len(only_in_1)}")
print(f"➖ Codes only in collectie: {len(only_in_2)}")

# === Venn Diagram ===
plt.figure(figsize=(8, 6))
venn2(subsets=(len(only_in_1), len(only_in_2), len(common_codes)),
      set_labels=('collection', 'collectie'))
plt.title("Overlap of Artwork Codes in 'collection' vs 'collectie'")
plt.tight_layout()

# === Save & show plot ===
venn_output = os.path.join(base_dir, 'collection_vs_collectie_venn.png')
plt.savefig(venn_output)
plt.show()

print(f"\n🖼️ Venn diagram saved to: {venn_output}")
print("✅ Comparison and visualization completed.")
```

### Extracting object numbers from IMAGE urls

```python
# Extracting codes from IMAGE urls

import pandas as pd
import re
import os

# Define file path
file_path = r'data/Rijksmuseum_urls_image_formats.csv'
base_dir = os.path.dirname(file_path)

# Updated regex pattern: match code before file suffix like .T.jpg or _F.jpg
code_pattern = re.compile(r'\b((?:[A-Z0-9]+-){1,}[A-Z0-9]+)(?=[._-]?[A-Z]?\.[a-z]{3,4})?')

# Function to extract code from URL
def extract_code(url):
    match = code_pattern.search(str(url))
    return match.group(1) if match else None

# Load the CSV and extract codes
df = pd.read_csv(file_path)
df['code'] = df['original'].apply(extract_code)

# Split into two DataFrames
df_with_code = df.dropna(subset=['code'])
df_without_code = df[df['code'].isna()]

# Save both DataFrames to CSV
with_code_path = os.path.join(base_dir, 'Rijksmuseum_IMAGE_urls_with_codes.csv')
without_code_path = os.path.join(base_dir, 'Rijksmuseum_IMAGE_urls_without_codes.csv')

df_with_code.to_csv(with_code_path, index=False)
df_without_code.to_csv(without_code_path, index=False)

# Print summary
print("✅ Extraction completed and CSV files saved:")
print(f"📄 Saved with codes: {with_code_path}")
print(f"📄 Saved without codes: {without_code_path}")
print(f"🔢 Total processed: {len(df)}")
print(f"🔑 With code: {len(df_with_code)}")
print(f"🚫 Without code: {len(df_without_code)}")

```

### Combining object numbers, creating a single file for the IA with the object numbers

```python
import pandas as pd
from pathlib import Path

# Base folder for input and output
base_dir = Path(r'data')

# Input files
file_names = [
    "codes_in_both.csv",
    "codes_only_in_collection.csv",
    "codes_only_in_collectie.csv",
]

dfs = []
for name in file_names:
    path = base_dir / name
    if not path.is_file():
        raise FileNotFoundError(f"Expected file '{path}' but it was not found.")
    
    # Read 'code' explicitly as string
    df_temp = pd.read_csv(path, dtype={"code": "string"})
    if "code" not in df_temp.columns:
        raise ValueError(
            f"File '{name}' does not contain a 'code' column. "
            f"Columns found: {list(df_temp.columns)}"
        )
    
    # Keep only the 'code' column, ensured as string
    df_temp["code"] = df_temp["code"].astype("string")
    dfs.append(df_temp[["code"]])

# Combine and deduplicate
combined = pd.concat(dfs, ignore_index=True)

# Drop rows where 'code' is missing, just in case
combined = combined.dropna(subset=["code"])

unique_codes = (
    combined
    .drop_duplicates(subset="code")
    .sort_values("code")
    .reset_index(drop=True)
)

# Ensure final column is string dtype
unique_codes["code"] = unique_codes["code"].astype("string")

# Output file (one column: 'code', as text)
output_path = base_dir / "object_numbers_IA.csv"
unique_codes.to_csv(output_path, index=False)

# Optional: check dtype and preview
print(unique_codes.dtypes)
unique_codes.head()
```

The resulting file has been manually checked and cleaned to remove obvious deviations such as numbers only strings
